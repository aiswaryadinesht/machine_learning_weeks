```python
import cv2
from datetime import datetime
cap=cv2.VideoCapture(0)
while True:
    ret,frame=cap.read()
    if not ret:
        break

    dt=datetime.now().strftime('%Y-%m-%d %H:%M:%S')
    cv2.putText(frame,dt,(10,50),cv2.FONT_HERSHEY_SIMPLEX,2,(0,255,0),2)

    cv2.imshow("camera",frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2

# Step 1: Load and display an image
image = cv2.imread('image.jpeg')  # Replace 'sample.jpg' with the path to your image
cv2.imshow('Original Image', image)

# Step 2: Resize the image
resized_image = cv2.resize(image, (300, 300))  # Resize to 300x300 pixels
cv2.imshow('Resized Image', resized_image)

# Step 3: Rotate the image
(h, w) = image.shape[:2]  # Get the height and width of the image
center = (w // 2, h // 2)  # Calculate the center of the image

# Rotate the image by 45 degrees
rotation_matrix = cv2.getRotationMatrix2D(center, 45, 1.0)  # Rotate around the center
rotated_image = cv2.warpAffine(image, rotation_matrix, (w, h))
cv2.imshow('Rotated Image', rotated_image)

# Step 4: Convert the image to grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2.imshow('Grayscale Image', gray_image)

# Step 5: Draw shapes on the image
# Create a copy of the original image to draw on
image_with_shapes = image.copy()

# Draw a red rectangle
cv2.rectangle(image_with_shapes, (50, 50), (200, 200), (0, 0, 255), 3)

# Draw a blue circle
cv2.circle(image_with_shapes, (300, 300), 50, (255, 0, 0), -1)  # -1 fills the circle

# Draw a green line
cv2.line(image_with_shapes, (0, 0), (w, h), (0, 255, 0), 2)

# Step 6: Add text to the image
font = cv2.FONT_HERSHEY_SIMPLEX
cv2.putText(image_with_shapes, 'OpenCV Demo', (50, 400), font, 1, (255, 255, 255), 2)

cv2.imshow('Image with Shapes and Text', image_with_shapes)

# Step 7: Wait for a key press and close all windows
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

# Step 1: Load the image
image = cv2.imread('image.jpeg', cv2.IMREAD_COLOR)  # Replace 'sample.jpg' with your image path
cv2.imshow('Original Image', image)

# Step 2: Apply different types of blurring

# 2.1. Average Blurring
average_blur = cv2.blur(image, (5, 5))  # Kernel size 5x5
cv2.imshow('Average Blur', average_blur)

# 2.2. Gaussian Blurring
gaussian_blur = cv2.GaussianBlur(image, (5, 5), 0)  # Kernel size 5x5, sigma = 0
cv2.imshow('Gaussian Blur', gaussian_blur)

# 2.3. Median Blurring
median_blur = cv2.medianBlur(image, 5)  # Kernel size 5
cv2.imshow('Median Blur', median_blur)

# 2.4. Bilateral Filtering (preserves edges)
bilateral_blur = cv2.bilateralFilter(image, 9, 75, 75)  # d=9, sigmaColor=75, sigmaSpace=75
cv2.imshow('Bilateral Blur', bilateral_blur)

# Step 3: Apply different types of thresholding

# Convert to grayscale for thresholding
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2.imshow('Grayscale Image', gray_image)

# 3.1. Simple Thresholding
_, simple_thresh = cv2.threshold(gray_image, 127, 255, cv2.THRESH_BINARY)  # Threshold value = 127
cv2.imshow('Simple Thresholding', simple_thresh)

# 3.2. Adaptive Thresholding
adaptive_thresh = cv2.adaptiveThreshold(
    gray_image, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2
)  # Block size = 11, C = 2
cv2.imshow('Adaptive Thresholding', adaptive_thresh)

# 3.3. Otsu's Thresholding
_, otsu_thresh = cv2.threshold(
    gray_image, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU
)
cv2.imshow("Otsu's Thresholding", otsu_thresh)

# Step 4: Edge Detection using Canny Edge Detector
# Apply Canny on grayscale image
edges = cv2.Canny(gray_image, 100, 200)  # Min and Max threshold for edges
cv2.imshow('Canny Edge Detection', edges)

# Step 5: Wait for key press and close all windows
cv2.waitKey(0)
cv2.destroyAllWindows()
```
