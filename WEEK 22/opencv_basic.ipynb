# Image Basics

```python
import cv2
image=cv2.imread("image.jpeg")
cv2.imshow("load image",image)

# Wait for a key press and close
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#to save the image
cv2.imwrite("output_image.jpg", image)
```

### Image Properties


```python
print("Shape:", image.shape)  # (height, width, channels)
print("Size:", image.size)    # Total number of pixels
print("Datatype:", image.dtype)  # Data type of pixel values
```

## Basic Operations on Images

```python
resized_image = cv2.resize(image, (200, 200))  # New dimensions: 300x300
cv2.imshow("Resized Image", resized_image)
cv2.waitKey(0)
```

```python
cropped_image = image[0:200, 100:300]  # Crop rows 50-200 and cols 100-300
cv2.imshow("Cropped Image", cropped_image)
cv2.waitKey(0)
```

```python
rotated_image = cv2.rotate(image, cv2.ROTATE_180)
cv2.imshow("Rotated Image", rotated_image)
cv2.waitKey(0)
```

```python
flipped_image = cv2.flip(image, 0)
cv2.imshow("Flipped Image", flipped_image)
cv2.waitKey(0)
```

# Drawing on Images

```python
import cv2

# Create a blank image
image = cv2.imread("image.jpeg")

# Draw a line: (start_x, start_y), (end_x, end_y), color, thickness
cv2.line(image, (50, 50), (200, 50), (0, 255, 0), 3)

# Draw a rectangle: (top-left), (bottom-right), color, thickness
cv2.rectangle(image, (50, 100), (200, 200), (255, 0, 0), 2)

# Draw a circle: (center_x, center_y), radius, color, thickness
cv2.circle(image, (300, 300), 50, (0, 0, 255), -1)  # Fill with -1

# Add Text
cv2.putText(image, "OpenCV Demo", (50, 400), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 255), 2)

cv2.imshow("Drawing on Image", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2.imshow("Grayscale Image", gray_image)
cv2.waitKey(0)
```

```python

hsv_image = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
lab_image = cv2.cvtColor(image, cv2.COLOR_BGR2LAB)

cv2.imshow("HSV Image", hsv_image)
cv2.imshow("LAB Image", lab_image)
cv2.waitKey(0)
```

```python
import cv2

# Load an image
image = cv2.imread("image.jpeg")

# Add text to the image
cv2.putText(image, "OpenCV Demo", (50, 400), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)

# Display the result
cv2.imshow("Image with Text", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

# Read an image in color
image = cv2.imread("image.jpeg", cv2.IMREAD_COLOR)

# Read an image in grayscale
gray_image = cv2.imread("image.jpeg", cv2.IMREAD_GRAYSCALE)

# Display images
cv2.imshow("Color Image", image)
cv2.imshow("Grayscale Image", gray_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
