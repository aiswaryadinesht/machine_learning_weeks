```python
import cv2
import numpy as np

# Create a black image
image = np.zeros((500, 500, 3), dtype="uint8")

# Draw a line
cv2.line(image, (50, 50), (400, 50), (255, 0, 0), 3)

# Draw a rectangle
cv2.rectangle(image, (50, 100), (200, 200), (0, 255, 0), -1)

# Draw a circle
cv2.circle(image, (250, 350), 100, (0, 255, 255), 5)

# Show the result
cv2.imshow('Shapes', image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
from datetime import datetime

# Open webcam
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break
    
    # Get current time
    current_time = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    
    # Add date and time text to frame
    cv2.putText(frame, current_time, (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 
                1, (0, 255, 0), 2, cv2.LINE_AA)
    
    cv2.imshow('Video with Time', frame)
    
    if cv2.waitKey(1) & 0xFF == 27:  # Press ESC to exit
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
# Define the mouse callback function
def mouse_event(event, x, y, flags, param):
    image1=np.zeros((1000,1000,3),dtype="uint8")
    if event == cv2.EVENT_LBUTTONDOWN:
       cv2.imshow("show",image1)
    elif event == cv2.EVENT_RBUTTONDOWN:
        print(f"Right Button Click at ({x}, {y})")
    elif event == cv2.EVENT_MOUSEMOVE:
        print(f"Mouse moved to ({x}, {y})")

# Create a black image
image = cv2.imread('image.jpeg')

# Set mouse callback
cv2.namedWindow('Mouse Event')
cv2.setMouseCallback('Mouse Event', mouse_event)

while True:
    cv2.imshow('Mouse Event', image)
    if cv2.waitKey(1) & 0xFF == 27:  # Press ESC to exit
        break

cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# Define the mouse callback function
def draw_shape(event, x, y, flags, param):
    global image
    if event == cv2.EVENT_LBUTTONDOWN:
        # Draw a circle on left button click
        cv2.circle(image, (x, y), 20, (255, 0, 0), -1)
    elif event == cv2.EVENT_RBUTTONDOWN:
        # Draw a rectangle on right button click
        cv2.rectangle(image, (x - 100, y - 20), (x + 100, y + 20), (0, 255, 0), -1)

# Create a black image
image = np.zeros((500, 500, 3), dtype="uint8")

# Set mouse callback
cv2.namedWindow('Drawing App')
cv2.setMouseCallback('Drawing App', draw_shape)

while True:
    cv2.imshow('Drawing App', image)
    if cv2.waitKey(1) & 0xFF == 27:  # Press ESC to exit
        break

cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
# Initialize global variables
drawing = False  # True if the mouse is pressed
start_x, start_y = -1, -1

# Define the mouse callback function
def draw_rectangle(event, x, y, flags, param):
    global start_x, start_y, drawing, image

    if event == cv2.EVENT_LBUTTONDOWN:
        # Start drawing (mouse pressed)
        drawing = True
        start_x, start_y = x, y

    elif event == cv2.EVENT_MOUSEMOVE:
        if drawing:  # If the mouse is pressed, update the rectangle
            temp_image = image.copy()
            cv2.rectangle(temp_image, (start_x, start_y), (x, y), (0, 255, 0), 2)
            cv2.imshow('Interactive Drawing', temp_image)

    elif event == cv2.EVENT_LBUTTONUP:
        # Finish drawing (mouse released)
        drawing = False
        cv2.rectangle(image, (start_x, start_y), (x, y), (0, 255, 0), 2)

# Create a blank image
#image = cv2.imread('sample.jpg')
#image=np.zeros((1000,1000,3),dtype="uint8")

# Set mouse callback
cv2.namedWindow('Interactive Drawing')
cv2.setMouseCallback('Interactive Drawing', draw_rectangle)

while True:
    cv2.imshow('Interactive Drawing', image)
    if cv2.waitKey(1) & 0xFF == 27:  # Press ESC to exit
        break

cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
# Mouse callback function with additional parameter
def mouse_event_with_param(event, x, y, flags, param):
    image = param
    if event == cv2.EVENT_LBUTTONDOWN:
        # Mark the clicked point with a circle
        cv2.circle(image, (x, y), 10, (0, 255, 0), 1)
        cv2.imshow("Mouse Callback with Param", image)

# Create an image and window
image=np.zeros((1000,1000,3),dtype="uint8")
# image = cv2.imread('sample.jpg')
cv2.namedWindow("Mouse Callback with Param")
cv2.setMouseCallback("Mouse Callback with Param", mouse_event_with_param, param=image)

while True:
    cv2.imshow("Mouse Callback with Param", image)
    if cv2.waitKey(1) & 0xFF == 27:  # Press ESC to exit
        break

cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# Global variables to track starting point
x_start, y_start = -1, -1
drawing = False  # Flag to check if drawing is in progress

# Mouse callback function
def mouse_event(event, x, y, flags, param):
    global x_start, y_start, drawing, image
    
    if event == cv2.EVENT_LBUTTONDOWN:  # Start drawing
        drawing = True
        x_start, y_start = x, y

    elif event == cv2.EVENT_MOUSEMOVE:  # While moving, draw dynamically if mouse is pressed
        if drawing:
            temp_image = image.copy()
            cv2.line(temp_image, (x_start, y_start), (x, y), (0, 255, 255), 2)
            cv2.imshow("test window", temp_image)

    elif event == cv2.EVENT_LBUTTONUP:  # Finish drawing
        drawing = False
        cv2.line(image, (x_start, y_start), (x, y), (0, 255, 255), 2)
        cv2.imshow("test window", image)

# Create a black canvas
image = np.zeros((500, 500, 3), dtype="uint8")  # 3 channels for RGB

# Create a window and bind the mouse callback
cv2.namedWindow("test window")
cv2.setMouseCallback("test window", mouse_event)

while True:
    cv2.imshow("test window", image)
    if cv2.waitKey(1) & 0xFF == 27:  # Press ESC to exit
        break

cv2.destroyAllWindows()
```
