```python
import cv2

# Read an image
image = cv2.imread('image.jpeg')  # Load an RGB image

# Split the image into Blue, Green, and Red channels
b, g, r = cv2.split(image)
# Resize each channel individually
b_resized = cv2.resize(b, (100, 100))
g_resized = cv2.resize(g, (100, 100))
r_resized = cv2.resize(r, (100, 100))
# Display the individual channels
cv2.imshow("Blue Channel", b_resized)
cv2.imshow("Green Channel", g_resized)
cv2.imshow("Red Channel", r_resized)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

# Read an image
image = cv2.imread('image.jpeg')

# Split the image into channels
b, g, r = cv2.split(image)

# Merge channels back into an RGB image
merged_image = cv2.merge((b, g, r))

# Display the original and merged images
cv2.imshow("Original Image", image)
cv2.imshow("Merged Image", merged_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

def on_trackbar(val):
    """Callback function for trackbar."""
    brightness = val - 100  # Adjusting brightness range (-100 to 100)
    modified_image = cv2.convertScaleAbs(image, alpha=1, beta=brightness)
    cv2.imshow('Image with Trackbar', modified_image)

# Load an image
image = cv2.imread('image.jpeg')

# Create a window
cv2.namedWindow('Image with Trackbar')

# Create a trackbar for brightness adjustment
cv2.createTrackbar('Brightness', 'Image with Trackbar', 100, 200, on_trackbar)

# Display the original image
cv2.imshow('Image with Trackbar', image)

cv2.waitKey(0)
cv2.destroyAllWindows()

```
