```python
import tensorflow as tf
import numpy as np

# Define matrices
matrix1 = np.array([(2, 2, 2), (2, 2, 2), (2, 2, 2)], dtype='int32')
matrix2 = np.array([(1, 1, 1), (1, 1, 1), (1, 1, 1)], dtype='int32')

print("Matrix 1:")
print(matrix1)
print("Matrix 2:")
print(matrix2)

# Convert matrices to TensorFlow constants
matrix1 = tf.constant(matrix1)
matrix2 = tf.constant(matrix2)

# TensorFlow operations
matrix_product = tf.matmul(matrix1, matrix2)
matrix_sum = tf.add(matrix1, matrix2)

# Define a third matrix for determinant calculation
matrix_3 = np.array([(2, 7, 2), (1, 4, 2), (9, 0, 2)], dtype='float32')
print("Matrix 3:")
print(matrix_3)

# Calculate determinant
matrix_det = tf.linalg.det(matrix_3)

# Display results
print("\nMatrix Product:")
print(matrix_product.numpy())

print("\nMatrix Sum:")
print(matrix_sum.numpy())

print("\nDeterminant of Matrix 3:")
print(matrix_det.numpy())
```

```python
import tensorflow as tf

# Data
X = tf.constant([1.0, 2.0, 3.0, 4.0], dtype=tf.float32)
y = tf.constant([5.0, 8.0, 11.0, 14.0], dtype=tf.float32)

# Variables
W = tf.Variable(0.0)
b = tf.Variable(0.0)

# Model
def linear_model(X):
    return W * X + b

# Loss function (Mean Squared Error)
def loss_fn(y_true, y_pred):
    return tf.reduce_mean(tf.square(y_true - y_pred))

# Optimizer
optimizer = tf.optimizers.Adam(learning_rate=0.1)

# Training
for epoch in range(100):
    with tf.GradientTape() as tape:
        predictions = linear_model(X)
        loss = loss_fn(y, predictions)
    gradients = tape.gradient(loss, [W, b])
    optimizer.apply_gradients(zip(gradients, [W, b]))

    if epoch % 10 == 0:
        print(f"Epoch {epoch}: Loss = {loss.numpy()}, W = {W.numpy()}, b = {b.numpy()}")

print(f"Final Model: y = {W.numpy()}x + {b.numpy()}")
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import numpy as np

# Data
X = np.random.rand(1000, 2)
y = np.array([1 if x[0] + x[1] > 1 else 0 for x in X])

# Model
model = Sequential([
    Dense(16, activation='relu', input_shape=(2,)),
    Dense(8, activation='relu'),
    Dense(1, activation='sigmoid')
])

# Compile
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Train
model.fit(X, y, epochs=10, batch_size=32)

# Predict
print("Predictions:", model.predict(X[:5]))
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, LSTM#, CuDNNLSTM


mnist = tf.keras.datasets.mnist  # mnist is a dataset of 28x28 images of handwritten digits and their labels
(x_train, y_train),(x_test, y_test) = mnist.load_data()  # unpacks images to x_train/x_test and labels to y_train/y_test

x_train = x_train/255.0
x_test = x_test/255.0

print(x_train.shape)
print(x_train[0].shape)

model = Sequential()

# IF you are running with a GPU, try out the CuDNNLSTM layer type instead (don't pass an activation, tanh is required)
model.add(LSTM(128, input_shape=(x_train.shape[1:]), activation='relu', return_sequences=True))
model.add(Dropout(0.2))

model.add(LSTM(128, activation='relu'))
model.add(Dropout(0.1))

model.add(Dense(32, activation='relu'))
model.add(Dropout(0.2))

model.add(Dense(10, activation='softmax'))

opt = tf.keras.optimizers.Adam(learning_rate=0.001, decay=1e-6)

# Compile model
model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer=opt,
    metrics=['accuracy'],
)

model.fit(x_train,
          y_train,
          epochs=3,
          validation_data=(x_test, y_test))
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, SimpleRNN

# Load MNIST dataset
mnist = tf.keras.datasets.mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalize the data
x_train = x_train / 255.0
x_test = x_test / 255.0

print(x_train.shape)
print(x_train[0].shape)

# Build the RNN model
model = Sequential()

# Add a SimpleRNN layer
model.add(SimpleRNN(128, input_shape=(x_train.shape[1:]), activation='relu', return_sequences=True))
model.add(Dropout(0.2))

# Add another SimpleRNN layer
model.add(SimpleRNN(128, activation='relu'))
model.add(Dropout(0.1))

# Add a fully connected Dense layer
model.add(Dense(32, activation='relu'))
model.add(Dropout(0.2))

# Output layer with 10 units (one for each digit, 0-9)
model.add(Dense(10, activation='softmax'))

# Compile the model
opt = tf.keras.optimizers.Adam(learning_rate=0.001, decay=1e-6)

model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer=opt,
    metrics=['accuracy'],
)

# Train the model
model.fit(x_train, y_train, epochs=3, validation_data=(x_test, y_test))
```
